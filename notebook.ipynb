```python
import cv2
import pytesseract
from PIL import Image

# 1. Load the image
image_path = "example.jpg"
image = cv2.imread(image_path)

# 2. Preprocessing (Crucial for book pages to remove shadows/creases)
# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Apply adaptive thresholding to create a clean black-and-white image
processed_img = cv2.adaptiveThreshold(
    gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2
)

# 3. Text Extraction using Tesseract
# --psm 3 tells Tesseract to look for a fully automatic page segmentation (good for books)
custom_config = r'--oem 3 --psm 3'
extracted_text = pytesseract.image_to_string(processed_img, config=custom_config)

# Print the final result
print(extracted_text)
```
