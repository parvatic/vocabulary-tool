```python
import cv2
import pytesseract
from PIL import Image

# 1. Load the image
image_path = "example.jpg"
image = cv2.imread(image_path)

# 2. Preprocessing
# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Upscale: the photo is small (~390x520 px) and Tesseract needs letters ~30 px tall.
# On such a small image, adaptive thresholding erases the text (Tesseract then returns "").
processed_img = cv2.resize(gray, None, fx=3, fy=3, interpolation=cv2.INTER_CUBIC)

# 3. Text Extraction using Tesseract
# --psm 3 tells Tesseract to look for a fully automatic page segmentation (good for books)
# lang="fra": the book is in French (accents, punctuation)
custom_config = r'--oem 3 --psm 3'
extracted_text = pytesseract.image_to_string(processed_img, lang="fra", config=custom_config)

# Print the final result
print(f"Text is {extracted_text}")
```
